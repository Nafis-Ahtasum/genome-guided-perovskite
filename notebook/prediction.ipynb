```python
df['Tolerance_Factor'] = (df['R-A(pm)'] + df['R-X(pm)']) / (np.sqrt(2) * ((df['R-B1(pm)'] + df['R-B3(pm)']) / 2 + df['R-X(pm)']))
```

```python
df['Octahedral_Factor'] = (df['R-B1(pm)'] + df['R-B3(pm)']) / (2 * df['R-X(pm)'])
```

```python
filtered_df = df[
    (df['band_gap'] >= 0.65) & (df['band_gap'] <= 1.6) &
    (df['Octahedral_Factor'] >= 0.44) & (df['Octahedral_Factor'] <= 0.9) &
    (df['Tolerance_Factor'] >= 0.82) & (df['Tolerance_Factor'] <= 1.08) &
    (df['stable'] == 1)
]
print(filtered_df.info())
```

```python
filtered_df[filtered_df['formula'] == 'Cs2SnMgBr6']
```

```python
import math
def calculate_new_tau(rx, rb, ra, na):
    try:
        if ra <= 0 or rb <= 0 or rx <= 0 or math.isclose(ra, rb):
            return np.nan
        term1 = rx / rb
        term2 = na * (na - (ra / rb) / math.log(ra / rb))
        tau = term1 - term2
        return tau
    except:
        return np.nan
```

```python
filtered_df['rb'] = (filtered_df['R-B1(pm)'] + filtered_df['R-B3(pm)']) / 2
na = 1
filtered_df['new_tau'] = filtered_df.apply(lambda row: calculate_new_tau(row['R-X(pm)'], row['rb'], row['R-A(pm)'], na), axis=1)
```

```python
final_df = filtered_df[filtered_df['new_tau'] < 4.18]
```
